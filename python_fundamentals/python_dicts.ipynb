### Python Dict

#### Counter
It is a specialized dictionary designed for counting hashable objects.

```python
from collections import Counter

# Counting characters in a string
word_count = Counter("kachapapadpakkapapad")
print(word_count)
```

In built methods

```python
#most_common(n): Returns a list of the n most frequent elements and their counts.

word_count.most_common(3)
```

```python
#elements(): Returns an iterator over elements, repeating each as many times as its count.
list(word_count.elements())
```

no key error unlike normal dict

```python
# existing key
word_count['k']
```

```python
# non exisitng key
word_count['z']
```

update the counter

```python
word_count.update('z')
```

```python
word_count['z']
```

```python
 

# Counting items in a list
inventory = Counter(['apple', 'banana', 'apple', 'orange', 'banana', 'apple'])
print(inventory)
# Output: Counter({'apple': 3, 'banana': 2, 'orange': 1})
```

```python
inventory.update(['kiwi'])
inventory
```

```python
# counter just matches the word if there increase the counter otherwise add it
inventory.update('kiwi')
inventory
```
