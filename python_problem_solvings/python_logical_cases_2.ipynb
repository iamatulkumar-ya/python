### Python - Logical Problems - Set 2

##### 1. Matrix Transposition: Achieve below output from the given metrix

Input:=> [[1,6,9],[6,7,1],[8,3,7]]

Output:=> [1,6,8], [6,7,3], [9,1,7]

Solution 1:

```python
li = [[1,6,9],[6,7,1],[8,3,7]]

di = {}
for i in li:
    for j, v in enumerate(i):
        if j in di:
            di[j].append(v)
        else:
            di[j] = [v]

list(di.values())
```

Solution 2:

```python
li = [[1,6,9],[6,7,1],[8,3,7]]
[list(item) for item in zip(*li)]  # *li , * will unpack each element and pass as separate argument
```

##### 2. Find nth largest number in given li 

Input:=> [1,8,4,9,45,67,32,74,68,24,15,21,69,6,7]

Output:=> Find 3rd largest = 68

Solution 1: Bubble short

```python
num_list = [1,8,4,9,45,67,32,74,68,24,15,21,69,6,7]
for i in range(len(num_list)):
    for j in range((len(num_list)-i-1)):
        if num_list[j] > num_list[j+1]:
            num_list[j], num_list[j+1] = num_list[j+1], num_list[j]

num_list[-3]
```

Solution 2: in-built method

```python
num_list = [1,8,4,9,45,67,32,74,68,24,15,21,69,6,7]
num_list.sort() # using sort as it will sort within list
num_list[-3]
```

```python
num_list.sort(reverse=True) # using sort as it will sort within list
num_list[2]
```

Solution 3: Heapq

```python
num_list = [1,8,4,9,45,67,32,74,68,24,15,21,69,6,7]

import heapq

# let's heapify means create heapq
heapq.heapify(num_list)

# let's find the nth largest as list and last index will be required one
heapq.nlargest(3, num_list)[-1]
```
